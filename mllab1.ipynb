```python
import numpy as np
import pandas as pd
import random

dataset = [10, 20, 30, 40, 50]
print("Original List:", dataset)

dataset.remove(40)
print("List after removing 40:", dataset)

dataset.append(60)
print("List after adding 60:", dataset)

random_number = random.choice(dataset)
print("The randomly selected number is:", random_number)
print("The dataset remains:", dataset)
```
