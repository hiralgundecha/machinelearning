```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
# Independent variable: Years of Experience
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# Dependent variable: Salary (₹1000)
Y = np.array([25, 30, 35, 40, 45, 50])
def gradient_descent(X, Y, learning_rate=0.01, iterations=1000):
    m = 0
    c = 0
    n = len(Y)

    X = X.flatten()  # convert to 1D array

    for _ in range(iterations):
        Y_pred = m * X + c

        dm = (-2/n) * np.sum(X * (Y - Y_pred))
        dc = (-2/n) * np.sum(Y - Y_pred)

        m = m - learning_rate * dm
        c = c - learning_rate * dc

    return m, c
m, c = gradient_descent(X, Y)

print(f"Manual Gradient Descent Model:")
print(f"Y = {m:.2f}X + {c:.2f}")
experience = 7
pred_salary = m * experience + c
print(f"Predicted Salary for {experience} years: ₹{pred_salary:.2f}K")
model = LinearRegression()
model.fit(X, Y)

print("\nScikit-learn Model:")
print(f"Y = {model.coef_[0]:.2f}X + {model.intercept_:.2f}")

sk_pred = model.predict([[7]])
print(f"Scikit-learn Prediction for 7 years: ₹{sk_pred[0]:.2f}K")
plt.scatter(X, Y, color="red", label="Actual Data")
plt.plot(X, m * X.flatten() + c, color="blue", label="Gradient Descent Line")
plt.plot(X, model.predict(X), color="green", linestyle="--", label="Scikit-learn Line")

plt.xlabel("Years of Experience")
plt.ylabel("Salary (₹1000)")
plt.title("Salary Prediction using Simple Linear Regression")
plt.legend()
plt.show()
```
